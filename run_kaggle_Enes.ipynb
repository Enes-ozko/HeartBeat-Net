```python
import os
import re
import sys
import json
import shutil

ROOT = '/kaggle/working/HeartBeat-Net'
if os.path.exists(ROOT):
    os.system('cd {} && git pull'.format(ROOT))
else:
    os.system('git clone -b enes https://github.com/Enes-ozko/HeartBeat-Net.git ' + ROOT)
```

```python
def run(notebook, **settings):
    folder = os.path.join(ROOT, os.path.dirname(notebook))
    os.chdir(folder)
    sys.path.insert(0, folder)
    namespace = {}
    for cell in json.load(open(os.path.basename(notebook)))['cells']:
        code = ''.join(cell['source'])
        for name, value in settings.items():
            code = re.sub('^{} = .*$'.format(name), '{} = {!r}'.format(name, value), code, flags=re.M)
        exec(code, namespace)
    sys.path.remove(folder)
```

```python
run('baseline/mainSegmentationChallenge_Enes.ipynb')
```

```python
run('semi-supervised/semiSupervised_Enes.ipynb')
```

```python
run('semi-supervised/meanTeacher_Enes.ipynb', FINAL=False)
```
